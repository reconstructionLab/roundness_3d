```python
"""
generate_example_synthetics.ipynb
Author: Henry Yuan
Description: Generate example synthetic grains for Figure 1 and Figure 3. Save .stl files
"""

# imports
import sys
sys.path.append('../../src/')
import synthetic_grains as sg
import os

# directory to save stls
save_dir = '../../data/synthetic/example_synthetics_fig1/'
os.makedirs(save_dir, exist_ok=True)

# parameters for synthetics, 9 total
aspects = [1,2,3]
rs=[0,0.5,1]
target_nv = 5000

N_ROUND = len(rs)
N_ASPECT = len(aspects)

# define rng
for i in range(N_ROUND):
    r = rs[i]
    print(f"Starting r = {r} synthetics...")
    for j in range(N_ASPECT):
        global_ind = i*N_ASPECT + j

        aspect = aspects[j]

        bv, mesh, nv = sg.synthetic_zircon(target_nv,aspect, r, rotate=False) # don't rotate these grains

        if nv >= 10000 or nv < 1000:
            print(nv)

        # Export to STL by index
        mesh.export(os.path.join(save_dir, f'mesh_{global_ind}.stl'))
    
```
